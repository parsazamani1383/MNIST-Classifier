# 🧠 **MNIST Digit Classification**

## 🎯 **Overview**
In this notebook, we build a **Machine Learning classifier (مدل دسته‌بندی)** to recognize handwritten digits using the **MNIST dataset**.

MNIST یکی از معروف‌ترین دیتاست‌های دنیای یادگیری ماشینه و برای تمرین و مقایسه مدل‌ها استفاده میشه.

---

## 📊 **Dataset**
**MNIST** شامل حدود **70,000 تصویر سیاه‌وسفید** از عددهای دست‌نویس بین **0 تا 9** هست.

- Image size (اندازه تصویر): `28x28`
- Total features (ویژگی‌ها): `784`
- Classes (کلاس‌ها): `10 digits (۰ تا ۹)`
- Pixel values (مقادیر پیکسل): از `0` (مشکی) تا `255` (سفید)

---

## 🎯 **Objective (هدف)**
قراره یه **classifier (مدل دسته‌بندی)** بسازیم که تشخیص بده هر تصویر عدد چندمه.

---

## ⚙️ **Workflow (روند کار)**
1. Load Dataset (لود دیتاست)  
2. Convert to DataFrame (تبدیل به دیتافریم)  
3. Data Exploration (بررسی داده)  
4. Train/Test Split (تفکیک داده آموزش و تست)  
5. Train Model (آموزش مدل)  
6. Evaluate Model (ارزیابی مدل)

---


```python
# 🏗️ Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Step 2: Load the MNIST dataset from OpenML
mnist = fetch_openml("mnist_784", version=1, as_frame=False)

# Extract features (X) and labels (y)
X = mnist.data
y = mnist.target

# Step 3: Convert the NumPy arrays into a Pandas DataFrame
# Create column names: pixel0, pixel1, ..., pixel783
pixel_columns = [f"pixel{i}" for i in range(X.shape[1])]

df = pd.DataFrame(X, columns=pixel_columns)
df["label"] = y  # Add target column

# Step 4: Show the first 5 rows to verify the structure
df.head()
```

# Visualizing Sample Images from MNIST

* Each row in the dataset represents a flattened image of size 28×28 pixels (784 features).

* To better understand the dataset, we visualize a few samples.

* We reshape each row back into a 28×28 image and display multiple digits in a 3×3 grid.

```python
# Visualize 9 sample images from the dataset
plt.figure(figsize=(6,6))  # size of the figure

for i in range(9):
    
    # take one flattened image (784 values)
    img = X[i]
    
    # reshape it back to 28x28 pixels
    img = img.reshape(28,28)

    # create subplot in a 3x3 grid
    plt.subplot(3,3,i+1)
    
    # display image
    plt.imshow(img, cmap="gray")
    
    # show the label of the digit
    plt.title(f"Label: {y[i]}")
    
    # remove axis for cleaner look
    plt.axis("off")

plt.tight_layout()
plt.show()
```

# Analyzing Class Distribution

* In a classification problem, it is important to understand how the data is distributed across classes.
* An imbalanced dataset can negatively affect the performance of machine learning models.
* Here, we count how many samples exist for each digit (0–9) in the MNIST dataset.
* We also visualize the distribution using a bar chart.


```python
# Analyze class distribution in the dataset

import pandas as pd
import matplotlib.pyplot as plt

# Convert labels to pandas Series for easier analysis
labels = pd.Series(y)

# Count number of samples for each digit
class_counts = labels.value_counts().sort_index()

# Print counts
print(class_counts)

# Plot distribution
plt.figure(figsize=(8,5))
plt.bar(class_counts.index, class_counts.values)

plt.xlabel("Digit")
plt.ylabel("Number of Samples")
plt.title("Class Distribution in MNIST")

plt.show()
```

# Splitting the Dataset into Training and Testing Sets

* In machine learning, we split the dataset into two parts: training set and testing set.
* The training set is used to train the model.
* The testing set is used to evaluate how well the model performs on unseen data.
* Here, we split the MNIST dataset using an 80/20 ratio.


```python
# Split dataset into training and testing sets

from sklearn.model_selection import train_test_split

# Perform the split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,      # 20% for testing
    random_state=42,    # ensures reproducibility
    stratify=y          # keeps class distribution balanced
)

# Print shapes of the resulting datasets
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

# Feature Scaling (Normalizing Pixel Values)

* The pixel values in the MNIST dataset range from 0 to 255.
* Many machine learning models perform better when input values are normalized.
* Here, we scale the pixel values to the range 0–1 by dividing by 255.
* This makes training more stable and can improve model performance.


```python
# Normalize pixel values to the range 0-1

# Convert to float (important for division)
X_train = X_train.astype("float32")
X_test  = X_test.astype("float32")

# Scale values
X_train /= 255.0
X_test  /= 255.0

# Check min/max values
print("Train min/max:", X_train.min(), X_train.max())
print("Test min/max:",  X_test.min(),  X_test.max())
```

# Training a Baseline Model (Logistic Regression)

* Before trying complex models, it is useful to train a simple baseline model.
* Logistic Regression is a common choice for evaluating initial performance on MNIST.
* This helps us understand how well a simple linear classifier performs.
* Later, we can compare more advanced models (e.g., neural networks) against this baseline.


```python
# Create the model
model = LogisticRegression(
    max_iter=1000,   # increase iterations for convergence
    solver="lbfgs",
    multi_class="multinomial",
    n_jobs=-1        # use all CPU cores
)

# Train the model
print("Training Logistic Regression...")
model.fit(X_train, y_train)

# Predict on test set
y_pred = model.predict(X_test)

# Evaluate accuracy
acc = accuracy_score(y_test, y_pred)

print("Test Accuracy:", acc)
```
